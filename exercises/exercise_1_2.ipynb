# Exercise 1.2: "Memory organization - C vs Fortran layout"
## Part A - theoretical:
### We have 6 elements stored contiguous in memory in the order: 1, 2, 3, 4, 5, 6.  In the following, we read this contiguous data into arrays in different ways.  What do the arrays look like if we read the data as:

    1. 2x3 matrix treating data as column-major (Fortran style) as F2x3?
        1 3 5
        2 4 6
    2. 3x2 matrix treating data as column-major (Fortran style) as F3x2?
        1 4
        2 5
        3 6
    3. 2x3 matrix treating data as row-major (C style) as C2x3?
        1 2 3
        4 5 6
    4. 3x2 matrix treating data as row-major (C style) as C3x2?
        1 2
        3 4
        5 6

### Explain the relations between the different matrices and how this may be utilized.
F2x3 is the transpose of C3x2
F3x2 is the transpose of C2x3

Choosing the right memory layout can significantly reduce cache misses and avoid unnecessary data copies.
